## Problem 1 — Two Sum (HashMap version)
**Importance:** Fastest solution for Two Sum.



### Concept
The Two Sum (HashMap version) problem gives an unsorted list and a fixed sum, so the goal is to report the pair of indices whose entries add up exactly to that target. Brute-force quadratic scans do not scale for interview sized inputs, so we derive the complement as we iterate and avoid revisiting earlier numbers.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Two Sum (HashMap version).

### Input
```
twoSum([2, 7, 11, 15], 9)
```

### Output
```
[0, 1]
```

### Reasoning
We walk through the array exactly once while keeping a hash map from value to index. At each step we check whether the current value's complement already lives in the map; if so we immediately return the stored index and the current index. Because we only ever store each index once, the operation stays linear in time while the map gives constant-time lookups.


````
# Problem 1: Two Sum (HashMap version) · Importance: Fastest solution for Two Sum.

def twoSum(nums, target):
    seen = {}
    for i, n in enumerate(nums):
        diff = target - n
        if diff in seen:
            return [seen[diff], i]
        seen[n] = i
    return []

print(twoSum([2, 7, 11, 15], 9))
````

**Importance:** Prefix map.



### Concept

### Reasoning
We first build the prefix structure (sum array, partial products, or parity) that encodes the contribution of every prefix. Then each query simply reads two entries and performs a constant-time subtraction or multiplication, so we do not recompute the same partial sums more than once. For cases that require two passes, we combine prefix and suffix runs to keep the complexity linear.


````
# Problem 6: Subarray Sum Equals K · Importance: Prefix map.

from collections import defaultdict

def subarraySum(nums, k):
    prefix = 0
    count = 0
    freq = defaultdict(int)
    freq[0] = 1
    for n in nums:
        prefix += n
        count += freq[prefix - k]
        freq[prefix] += 1
    return count

print(subarraySum([1,1,1], 2))
````

````
# Problem 7: Continuous Subarray Sum · Importance: Modulo prefix sum.

def checkSubarraySum(nums, k):
    seen = {0: -1}
    prefix = 0
    for i, n in enumerate(nums):
        prefix += n
        if k != 0:
            prefix %= k
        if prefix in seen:
            if i - seen[prefix] > 1:
                return True
        else:
            seen[prefix] = i
    return False

print(checkSubarraySum([23,2,4,6,7], 6))
````

## Problem 9 — LRU Cache
**Importance:** Dict + OrderedDict.



### Concept
LRU Cache demands the construction of a small stateful data structure (cache, random set, or time-aware map) that supports several operations with strict time constraints. The emphasis is on combining hash tables with ordered containers or lists to provide constant-time all operations.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to LRU Cache.

### Input
```

cache.put(2, 2)
cache.get(1)
```

### Output
```
1
-1
```

### Reasoning
We keep a hash map for direct lookups and either an ordered dict, list, or priority structure to maintain eviction or order semantics. Every method updates both structures in lockstep, so even though the data structure holds more than one piece of state, each public operation still completes in O(1) amortized time.


````
# Problem 9: LRU Cache · Importance: Dict + OrderedDict.

from collections import OrderedDict

class LRUCache(OrderedDict):
    def __init__(self, capacity):
        super().__init__()
        self.capacity = capacity

    def get(self, key):
        if key not in self:
            return -1
        self.move_to_end(key)
        return self[key]

    def put(self, key, value):
        if key in self:
            self.move_to_end(key)
        self[key] = value
        if len(self) > self.capacity:
            self.popitem(last=False)

cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print(cache.get(1))
cache.put(3, 3)
print(cache.get(2))
````

## Problem 10 — Insert Delete GetRandom O(1)
**Importance:** Dict + List trick.



### Concept
Insert Delete GetRandom O(1) demands the construction of a small stateful data structure (cache, random set, or time-aware map) that supports several operations with strict time constraints. The emphasis is on combining hash tables with ordered containers or lists to provide constant-time all operations.

### Description

### Input
```

obj.insert(1)
obj.insert(2)
obj.remove(1)
obj.getRandom()
```

### Output
```
2
```

### Reasoning
We keep a hash map for direct lookups and either an ordered dict, list, or priority structure to maintain eviction or order semantics. Every method updates both structures in lockstep, so even though the data structure holds more than one piece of state, each public operation still completes in O(1) amortized time.


````
# Problem 10: Insert Delete GetRandom O(1) · Importance: Dict + List trick.

import random

class RandomizedSet:
    def __init__(self):
        self.d = {}
        self.l = []

    def insert(self, val):
        if val in self.d:
            return False
        self.d[val] = len(self.l)
        self.l.append(val)
        return True

    def remove(self, val):
        if val not in self.d:
            return False
        idx, last = self.d[val], self.l[-1]
        self.l[idx], self.d[last] = last, idx
        self.l.pop()
        del self.d[val]
        return True

    def getRandom(self):
        return random.choice(self.l)

obj = RandomizedSet()
obj.insert(1)
obj.insert(2)
obj.remove(1)
print(obj.getRandom())
````

````
# Problem 13: Happy Number · Importance: Floyd or Hash Set.

def isHappy(n):
    def next_num(x):
        return sum(int(d) ** 2 for d in str(x))

    seen = set()
    while n not in seen:
        seen.add(n)
        n = next_num(n)
        if n == 1:
            return True
    return False

print(isHappy(19))
````

## Problem 14 — Contains Duplicate II
**Importance:** Sliding window hash set.



### Concept
Contains Duplicate II is solved by treating the input as a streaming sequence and expanding or contracting a window to satisfy the problem constraints. The window boundaries move forward while you aggregate counts, sums, or other metrics, which keeps the work linear even when multiple candidates exist.

### Description

### Input


### Reasoning
As we extend the right boundary we incorporate the newest value into our aggregate; if the current window violates the constraint we advance the left boundary until things become valid again. This approach tracks data in constant-time update operations per element, avoiding redundant scanning of the same prefix multiple times and yielding a predictable O(n) runtime.


````
# Problem 14: Contains Duplicate II · Importance: Sliding window hash set.

def containsNearbyDuplicate(nums, k):
    seen = {}
    for i, n in enumerate(nums):
        if n in seen and i - seen[n] <= k:
            return True
        seen[n] = i
    return False

print(containsNearbyDuplicate([1,0,1,1], 1))
````

````
# Problem 15: Word Pattern · Importance: Bijective mapping.

def wordPattern(pattern, s):
    words = s.split()
    if len(pattern) != len(words):
        return False
    p2w = {}
    w2p = {}
    for p, w in zip(pattern, words):
        if (p in p2w and p2w[p] != w) or (w in w2p and w2p[w] != p):
            return False
        p2w[p] = w
        w2p[w] = p
    return True

print(wordPattern("abba", "dog cat cat dog"))
````

## Problem 16 — Clone Graph
**Importance:** Node mapping.



### Concept
Clone Graph leverages relationships between nodes—for example, clone, trust, or angular dependencies—so you model the problem as a graph and traverse it with DFS/BFS or indegree counting.

### Description


### Output

### Reasoning
We either walk the graph recursively (DFS) or iterate through adjacency lists while tracking visited state, indegrees, or parent relationships. These traversals allow us to rebuild copies, find judges, or check trustworthiness within linear time in the number of nodes plus edges.


````
# Problem 16: Clone Graph · Importance: Node mapping.

class Node:
    def __init__(self, val, neighbors=None):
        self.val = val
        self.neighbors = neighbors or []

def cloneGraph(node):
    if not node:
        return None
    old_to_new = {}

    def dfs(cur):
        if cur in old_to_new:
            return old_to_new[cur]
        copy = Node(cur.val)
        old_to_new[cur] = copy
        for neighbor in cur.neighbors:
            copy.neighbors.append(dfs(neighbor))
        return copy

    return dfs(node)

# build tiny graph
node1 = Node(1)
node2 = Node(2)
node1.neighbors.append(node2)
node2.neighbors.append(node1)
copy = cloneGraph(node1)
print(copy.val, [n.val for n in copy.neighbors])
````

## Problem 17 — Find Town Judge
**Importance:** Trust mapping (Indegree/Outdegree).



### Concept
Find Town Judge leverages relationships between nodes—for example, clone, trust, or angular dependencies—so you model the problem as a graph and traverse it with DFS/BFS or indegree counting.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Find Town Judge.

### Input
```
findJudge(3, [[1,3],[2,3]])
```

### Output
```

```

### Reasoning
We either walk the graph recursively (DFS) or iterate through adjacency lists while tracking visited state, indegrees, or parent relationships. These traversals allow us to rebuild copies, find judges, or check trustworthiness within linear time in the number of nodes plus edges.


````
# Problem 17: Find Town Judge · Importance: Trust mapping (Indegree/Outdegree).

def findJudge(n, trust):
    indegree = [0] * (n + 1)
    outdegree = [0] * (n + 1)
    for a, b in trust:
        outdegree[a] += 1
        indegree[b] += 1
    for person in range(1, n + 1):
        if indegree[person] == n - 1 and outdegree[person] == 0:
            return person
    return -1

print(findJudge(3, [[1,3],[2,3]]))
````

## Problem 50 — Nearby Almost Duplicate
**Importance:** Value window



### Concept
Nearby Almost Duplicate asks you to maintain a window of values that must stay ordered (with an implicit BST or sorted container) so you can quickly find the nearest value that satisfies the distance constraint.

### Description


### Reasoning
We keep an ordered list of the current window using a sorted list or bisect operations. When we insert a new element, we search for its neighbors to verify the value constraint, and when the window grows too large we drop the oldest entry—all in logarithmic time per update.


````
# Problem 50: Nearby Almost Duplicate · Importance: Value window

def containsNearbyAlmostDuplicate(nums, k, t):
    import bisect
    window = []
    for i, n in enumerate(nums):
        pos = bisect.bisect_left(window, n)
        if pos < len(window) and abs(window[pos] - n) <= t:
            return True
        if pos > 0 and abs(window[pos-1] - n) <= t:
            return True
        window.insert(pos, n)
        if len(window) > k:
            window.pop(bisect.bisect_left(window, nums[i - k]))
    return False
print(containsNearbyAlmostDuplicate([1,5,9,1], 3, 0))
````
