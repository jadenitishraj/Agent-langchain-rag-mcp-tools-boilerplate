## Problem 1 — Two Sum
**Importance:** Entry to HashMaps. O(n) lookup.



### Concept
The Two Sum problem gives an unsorted list and a fixed sum, so the goal is to report the pair of indices whose entries add up exactly to that target. Brute-force quadratic scans do not scale for interview sized inputs, so we derive the complement as we iterate and avoid revisiting earlier numbers.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Two Sum.

### Input
```
twoSum([2, 7, 11, 15], 9)
```

### Output
```
[0, 1]
```

### Reasoning
We walk through the array exactly once while keeping a hash map from value to index. At each step we check whether the current value's complement already lives in the map; if so we immediately return the stored index and the current index. Because we only ever store each index once, the operation stays linear in time while the map gives constant-time lookups.


````python
# Problem 1: Two Sum · Importance: Entry to HashMaps. O(n) lookup.

def twoSum(nums, target):
    prevMap = {}
    for i, n in enumerate(nums):
        diff = target - n
        if diff in prevMap: return [prevMap[diff], i]
        prevMap[n] = i
print(twoSum([2, 7, 11, 15], 9))
````

## Problem 2 — Best Time to Buy and Sell Stock
**Importance:** Tracking min seen so far.



### Concept
Best Time to Buy and Sell Stock invites a greedy strategy: as you scan the array, keep track of the most promising candidate so far (minimum price, farthest reachable, or highest ascending streak) and update your best result whenever a better opportunity appears. The trick is to never reconsider older values if a better one has already been recorded.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Best Time to Buy and Sell Stock.

### Input
```
maxProfit([7,1,5,3,6,4])
```

### Output
```
5
```

### Reasoning
We maintain running variables such as minimum seen so far, current max gap, or the best window in process. When we encounter a new element that improves our state, we update the result; otherwise we slide past the element without undoing progress. Since each element is inspected once, this yields O(n) performance while respecting the greedy invariant that an earlier choice never needs to be reconsidered.


````python
# Problem 2: Best Time to Buy and Sell Stock · Importance: Tracking min seen so far.

def maxProfit(prices):
    l, r = 0, 1
    maxP = 0
    while r < len(prices):
        if prices[l] < prices[r]:
            maxP = max(maxP, prices[r] - prices[l])
        else: l = r
        r += 1
    return maxP
print(maxProfit([7,1,5,3,6,4]))
````

## Problem 3 — Maximum Subarray (Kadane)
**Importance:** Simplest DP/Greedy.



### Concept
Maximum Subarray (Kadane) requires tracking running minima or maxima, often maintaining both local and global values (e.g., current subarray sum, max product, longest product) as the sliding window evolves. The idea is to propagate state from one step to the next instead of recomputing from scratch.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Maximum Subarray (Kadane).

### Input
```
maxSubArray([-2,1,-3,4,-1,2,1,-5,4])
```

### Output
```
6
```

### Reasoning
We keep the best result seen so far as well as the best candidate ending at the current index. Because the next state depends only on the current element and the previous state, we can update in constant time while scanning the array. This captures the Kadane or max-product pattern piece by piece.


````python
# Problem 3: Maximum Subarray (Kadane) · Importance: Simplest DP/Greedy.

def maxSubArray(nums):
    maxSub = nums[0]
    curSum = 0
    for n in nums:
        curSum = max(n, curSum + n)
        maxSub = max(maxSub, curSum)
    return maxSub
print(maxSubArray([-2,1,-3,4,-1,2,1,-5,4]))
````

````python
# Problem 5: Product of Array Except Self · Importance: Prefix/Suffix products without division.

def productExceptSelf(nums):
    res = [1] * len(nums)
    prefix = 1
    for i in range(len(nums)):
        res[i] = prefix
        prefix *= nums[i]
    postfix = 1
    for i in range(len(nums) - 1, -1, -1):
        res[i] *= postfix
        postfix *= nums[i]
    return res
print(productExceptSelf([1,2,3,4]))
````

This captures the Kadane or max-product pattern piece by piece.


````python
# Problem 6: Maximum Product Subarray · Importance: Kadane variation with min/max.

def maxProduct(nums):
    res = max(nums)
    curMin, curMax = 1, 1
    for n in nums:
        tmp = curMax * n
        curMax = max(n * curMax, n * curMin, n)
        curMin = min(tmp, n * curMin, n)
        res = max(res, curMax)
    return res
print(maxProduct([2,3,-2,4]))
````

````python
# Problem 10: Container With Most Water · Importance: Greedy shrinkage.

def maxArea(height):
    l, r, res = 0, len(height)-1, 0
    while l < r:
        res = max(res, (r-l) * min(height[l], height[r]))
        if height[l] < height[r]: l += 1
        else: r -= 1
    return res
print(maxArea([1,8,6,2,5,4,8,3,7]))
````

## Problem 12 — Maximum Average Subarray
**Importance:** Fixed window foundation.



### Concept
Maximum Average Subarray is solved by treating the input as a streaming sequence and expanding or contracting a window to satisfy the problem constraints. The window boundaries move forward while you aggregate counts, sums, or other metrics, which keeps the work linear even when multiple candidates exist.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Maximum Average Subarray.

### Input
```
findMaxAverage([1,12,-5,-6,50,3], 4)
```

### Output
```
12.75
```

### Reasoning
As we extend the right boundary we incorporate the newest value into our aggregate; if the current window violates the constraint we advance the left boundary until things become valid again. This approach tracks data in constant-time update operations per element, avoiding redundant scanning of the same prefix multiple times and yielding a predictable O(n) runtime.


````python
# Problem 12: Maximum Average Subarray · Importance: Fixed window foundation.

def findMaxAverage(nums, k):
    curr = sum(nums[:k])
    res = curr
    for i in range(k, len(nums)):
        curr += nums[i] - nums[i-k]
        res = max(res, curr)
    return res / k
print(findMaxAverage([1,12,-5,-6,50,3], 4))
````

````python
# Problem 13: Longest Subarray of 1s · Importance: Variable window transformation.

def longestSubarray(nums):
    l, zeros, res = 0, 0, 0
    for r in range(len(nums)):
        zeros += (1 if nums[r] == 0 else 0)
        while zeros > 1:
            zeros -= (1 if nums[l] == 0 else 0)
            l += 1
        res = max(res, r - l)
    return res
print(longestSubarray([1,1,0,1]))
````

As we extend the right boundary we incorporate the newest value into our aggregate; if the current window violates the constraint we advance the left boundary until things become valid again.
````python
# Problem 14: Minimum Size Subarray Sum · Importance: O(n) greedy shrinkage.

def minSubArrayLen(target, nums):
    l, total, res = 0, 0, float('inf')
    for r in range(len(nums)):
        total += nums[r]
        while total >= target:
            res = min(res, r-l+1)
            total -= nums[l]
            l += 1
    return res if res != float('inf') else 0
print(minSubArrayLen(7, [2,3,1,2,4,3]))
````

## Problem 15 — Sliding Window Maximum
**Importance:** Monotonic Queue (Deque).



### Concept
Sliding Window Maximum is solved by treating the input as a streaming sequence and expanding or contracting a window to satisfy the problem constraints. The window boundaries move forward while you aggregate counts, sums, or other metrics, which keeps the work linear even when multiple candidates exist.

### Description

```
maxSlidingWindow([1,3,-1,-3,5,3,6,7], 3)
```

### Output
```
[3, 3, 5, 5, 6, 7]
```

### Reasoning
As we extend the right boundary we incorporate the newest value into our aggregate; if the current window violates the constraint we advance the left boundary until things become valid again. This approach tracks data in constant-time update operations per element, avoiding redundant scanning of the same prefix multiple times and yielding a predictable O(n) runtime.


````python
# Problem 15: Sliding Window Maximum · Importance: Monotonic Queue (Deque).

from collections import deque
def maxSlidingWindow(nums, k):
    q, res = deque(), []
    for i, n in enumerate(nums):
        while q and nums[q[-1]] < n: q.pop()
        q.append(i)
        if q[0] == i - k: q.popleft()
        if i >= k - 1: res.append(nums[q[0]])
    return res
print(maxSlidingWindow([1,3,-1,-3,5,3,6,7], 3))
````

````python
# Problem 19: Find Pivot Index · Importance: Equilibrium Point.

def pivotIndex(nums):
    S, left = sum(nums), 0
    for i, x in enumerate(nums):
        if left == (S - left - x): return i
        left += x
    return -1
print(pivotIndex([1,7,3,6,5,6]))
````

## Problem 45 — Maximum Ascending Sum
**Importance:** Stretch sum



### Concept
Maximum Ascending Sum invites a greedy strategy: as you scan the array, keep track of the most promising candidate so far (minimum price, farthest reachable, or highest ascending streak) and update your best result whenever a better opportunity appears. The trick is to never reconsider older values if a better one has already been recorded.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Maximum Ascending Sum.

### Input
```

```

### Output
```

```

### Reasoning
We maintain running variables such as minimum seen so far, current max gap, or the best window in process. When we encounter a new element that improves our state, we update the result; otherwise we slide past the element without undoing progress. Since each element is inspected once, this yields O(n) performance while respecting the greedy invariant that an earlier choice never needs to be reconsidered.


````python
# Problem 45: Maximum Ascending Sum · Importance: Stretch sum

def maxAscendingSum(nums):
    res = cur = nums[0]
    prev = nums[0]
    for n in nums[1:]:
        if n > prev:
            cur += n
        else:
            res = max(res, cur)
            cur = n
        prev = n
    return max(res, cur)
print(maxAscendingSum([10,20,30,5,10,50]))
````

As we move from left to right we push candidate indices onto the stack, and whenever a larger element appears we pop the stack and record the match. Each element enters and leaves the stack at most once, so the algorithm runs in linear time with only O(n) additional space.


````python
# Problem 51: Next Greater Element · Importance: Monotonic stack

def nextGreaterElement(nums):
    res = [-1] * len(nums)
    stack = []
    for i, n in enumerate(nums):
        while stack and nums[stack[-1]] < n:
            res[stack.pop()] = n
        stack.append(i)
    return res
print(nextGreaterElement([2,1,2,4,3]))
````

## Problem 59 — Nearby Duplicate

### Concept

As we extend the right boundary we incorporate the newest value into our aggregate; if the current window violates the constraint we advance the left boundary until things become valid again.
````python
# Problem 59: Nearby Duplicate · Importance: Index window

def containsNearbyDuplicate(nums, k):
    seen = {}
    for i, n in enumerate(nums):
        if n in seen and i - seen[n] <= k:
            return True
        seen[n] = i
    return False
print(containsNearbyDuplicate([1,2,3,1], 3))
````

## Problem 60 — Nearby Almost Duplicate
**Importance:** Value window



### Concept
Nearby Almost Duplicate asks you to maintain a window of values that must stay ordered (with an implicit BST or sorted container) so you can quickly find the nearest value that satisfies the distance constraint.

### Description
Apply the concept above to the sample data block shown below so that the output matches what the prompt expects. Scale the same logic to any valid input, respecting the constraints inherent to Nearby Almost Duplicate.

### Input
```
containsNearbyAlmostDuplicate([1,5,9,1], 3, 0)
```

### Output
```

```

### Reasoning
We keep an ordered list of the current window using a sorted list or bisect operations. When we insert a new element, we search for its neighbors to verify the value constraint, and when the window grows too large we drop the oldest entry—all in logarithmic time per update.


````python
# Problem 60: Nearby Almost Duplicate · Importance: Value window

def containsNearbyAlmostDuplicate(nums, k, t):
    import bisect
    window = []
    for i, n in enumerate(nums):
        pos = bisect.bisect_left(window, n)
        if pos < len(window) and abs(window[pos] - n) <= t:
            return True
        if pos > 0 and abs(window[pos-1] - n) <= t:
            return True
        window.insert(pos, n)
        if len(window) > k:
            window.pop(bisect.bisect_left(window, nums[i - k]))
    return False
print(containsNearbyAlmostDuplicate([1,5,9,1], 3, 0))
````
