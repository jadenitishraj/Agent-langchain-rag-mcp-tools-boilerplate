````
# Problem 6: Longest Substring Without Repeating Characters · Importance: Variable window.

def lengthOfLongestSubstring(s):
    start = 0
    seen = {}
    best = 0
    for i, ch in enumerate(s):
        if ch in seen and seen[ch] >= start:
            start = seen[ch] + 1
        seen[ch] = i
        best = max(best, i - start + 1)
    return best

print(lengthOfLongestSubstring("abcabcbb"))
````

## Problem 9 — Minimum Window Substring

### Input

```
minWindow("ADOBECODEBANC", "ABC")
```

### Output


### Reasoning
As we extend the right boundary we incorporate the newest value into our aggregate; if the current window violates the constraint we advance the left boundary until things become valid again. This approach tracks data in constant-time update operations per element, avoiding redundant scanning of the same prefix multiple times and yielding a predictable O(n) runtime.


````
# Problem 9: Minimum Window Substring · Importance: Two pointers + counts.

from collections import Counter

def minWindow(s, t):
    need = Counter(t)
    window = Counter()
    left = 0
    formed = 0
    required = len(need)
    best = (float('inf'), None, None)
    for right, ch in enumerate(s):
        window[ch] += 1
        if window[ch] == need[ch]:
            formed += 1
        while left <= right and formed == required:
            if right - left + 1 < best[0]:
                best = (right - left + 1, left, right)
            window[s[left]] -= 1
            if window[s[left]] < need[s[left]]:
                formed -= 1
            left += 1
    return '' if best[0] == float('inf') else s[best[1]:best[2]+1]

print(minWindow("ADOBECODEBANC", "ABC"))
````

````
# Problem 10: String to Integer (atoi) · Importance: Edge-case parsing.

def myAtoi(s):
    s = s.strip()
    if not s:
        return 0
    sign = 1
    idx = 0
    if s[0] in '+-':
        if s[0] == '-':
            sign = -1
        idx += 1
    res = 0
    while idx < len(s) and s[idx].isdigit():
        res = res * 10 + int(s[idx])
        idx += 1
    res *= sign
    return max(min(res, 2**31 - 1), -2**31)

print(myAtoi("   -42"))
````

````
# Problem 11: Roman to Integer · Importance: Mapping exercise.

def romanToInt(s):
    mapping = {'I':1,'V':5,'X':10,'L':50,'C':100,'D':500,'M':1000}
    res = 0
    prev = 0
    for ch in s[::-1]:
        if mapping[ch] < prev:
            res -= mapping[ch]
        else:
            res += mapping[ch]
        prev = mapping[ch]
    return res

print(romanToInt("MCMXCIV"))
````

````
# Problem 13: Longest Palindromic Substring · Importance: Expand around center.

def longestPalindrome(s):
    def expand(left, right):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return left + 1, right
    start, end = 0, 0
    for i in range(len(s)):
        l, r = expand(i, i)
        if r - l > end - start:
            start, end = l, r
        l, r = expand(i, i + 1)
        if r - l > end - start:
            start, end = l, r
    return s[start:end]

print(longestPalindrome("babad"))
````

````
# Problem 18: Zigzag Conversion · Importance: Stripe build.

def convert(s, numRows):
    if numRows == 1 or numRows >= len(s):
        return s
    rows = [''] * numRows
    cur_row, step = 0, 1
    for ch in s:
        rows[cur_row] += ch
        if cur_row == 0:
            step = 1
        elif cur_row == numRows - 1:
            step = -1
        cur_row += step
    return ''.join(rows)

print(convert("PAYPALISHIRING", 3))
````

````
# Problem 19: Multiply Strings · Importance: Manual product.

def multiply(num1, num2):
    if num1 == '0' or num2 == '0':
        return '0'
    res = [0] * (len(num1) + len(num2))
    for i in range(len(num1)-1, -1, -1):
        for j in range(len(num2)-1, -1, -1):
            res[i + j + 1] += int(num1[i]) * int(num2[j])
    for i in range(len(res)-1, 0, -1):
        carry = res[i] // 10
        res[i-1] += carry
        res[i] %= 10
    start = 0
    while start < len(res) and res[start] == 0:
        start += 1
    return ''.join(str(d) for d in res[start:])

print(multiply('123', '456'))
````

### Reasoning
We keep an ordered list of the current window using a sorted list or bisect operations. When we insert a new element, we search for its neighbors to verify the value constraint, and when the window grows too large we drop the oldest entry—all in logarithmic time per update.


````
# Problem 50: Nearby Almost Duplicate · Importance: Value window

def containsNearbyAlmostDuplicate(nums, k, t):
    import bisect
    window = []
    for i, n in enumerate(nums):
        pos = bisect.bisect_left(window, n)
        if pos < len(window) and abs(window[pos] - n) <= t:
            return True
        if pos > 0 and abs(window[pos-1] - n) <= t:
            return True
        window.insert(pos, n)
        if len(window) > k:
            window.pop(bisect.bisect_left(window, nums[i - k]))
    return False
print(containsNearbyAlmostDuplicate([1,5,9,1], 3, 0))
````

As we extend the right boundary we incorporate the newest value into our aggregate; if the current window violates the constraint we advance the left boundary until things become valid again.
````
# Problem 59: Longest Substring with At Most Two Distinct Characters · Importance: Sliding window.

from collections import defaultdict

def lengthOfLongestSubstringTwoDistinct(s):
    counts = defaultdict(int)
    left = 0
    max_len = 0
    for right, ch in enumerate(s):
        counts[ch] += 1
        while len(counts) > 2:
            counts[s[left]] -= 1
            if counts[s[left]] == 0:
                del counts[s[left]]
            left += 1
        max_len = max(max_len, right - left + 1)
    return max_len

print(lengthOfLongestSubstringTwoDistinct("eceba"))
````
